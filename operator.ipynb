Operators are special symbols in Python which are used to perform arithmetic and logical operations.

## Arithematic operators

```python
numberOne = 100
numberTwo = 10
```

```python
result = numberOne + numberTwo # Addition operator
print(result)
```

```python
result = numberOne - numberTwo # Subtraction operator
print(result)
```

```python
result = numberOne / numberTwo # Divison operator
print(result)
```

```python
result = numberOne % numberTwo # Modulus operator and stores remainder
print(result)
```

```python
numberOne = 2
numberTwo = 3
result = numberOne ** numberTwo # Exponent
print(result)
```

```python
numberOne = 15
numberTwo = 2
result = numberOne // numberTwo # Floor divison, rounds down in both case which also means "bigger" number in negative sense but is actually smaller like in positive sense
print(result)
```

## Assignment operator

Any operator which assigns value to the variable it maybe +=, //= or any with = it will be consider an assignment operator

```python
number = 10 # assignment operator
```

```python
x = 3
x >>= 1
print(x) # bitwise operation discards right most bit and shifts bit to the right step by step
```

```python
x = 3
x <<= 1
print(x) # bitwise operation discards left most bit and shifts bit to the left step by step
```

```python
x = 3
x ^= 1
print(x) # XOR
```

## Comparison operators

```python
x = 10
y = 20

print(x > y)   # x Greater than y
```

```python
x = 10
y = 20

print(x < y)   # x less than y
```

```python
x = 10
y = 10

print(x <= y)   # x less than equal y
```

```python
x = 11
y = 20

print(x > y)   # x greater than y
```

```python
x = 11
y = 20

print(x != y)   # x not equal to y
```

## Logical operators

```python
x = 5
print(x > 3 and x < 10) # AND operator, both conditon must be true.
```

```python
x = 5
print(x > 3 or x < 4) # OR operator, one condition must be true
```

```python
x = 10
print(x > 3 and x <= 10)
print(not(x > 3 and x <= 10)) # As demonstrated above the result should return true but adding not reverses the result
```

## Indentity operator
- is 
- not is

```python
x = ["fox", "cat"]
y = ["fox", "cat"]
z = x

print(x is z)
# returns True because z is the same object as x

print(x is y)
# returns False because y is a different object than x
```

## Membership operator
- in 
- not in

```python
search = "cat"
list = ["dog", "cat", "bird"]
print(search in list) # returns True because it is in the list
```

in is as effective as the data structure, if using list it will give time complexity of O(n) and incase of dictionary or set it will give O(1).

## Order of precedence
- Parenthese 
- Exponent 
- Multiplication and Division (from left to right)
- Addition and Subtraction (from left to right)
- Bitwise operators (from left to right)
- Logical operators (AND, OR, NOT)
- Identity and Membership operators (is, not in)

```python
print(5 + 4 - 7 + 3) # Since + and - have same precedence it will start from left to right
```

```python
print((6 + 3) - (6 + 3))
```
