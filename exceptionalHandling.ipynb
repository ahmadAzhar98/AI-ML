Control strcutures are blocks which analyze variables and chooses the direction program takes based on parameters.

```python
number = 10 # If statement

if number > 5:
    print("Number is greater than 5")
```

```python
number = 10 # If statement

if number < 5:
    print("Number is less than 5")
else:
    print("Number is greater than 5")
```

```python

#incase user divides by zero
try:
    dividend = 100
    divisor = 0
    quotient = dividend / divisor
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print(f"Quotient is {quotient}")
finally:
    print("This code will always run")                
print("Hello")    
```

Like in the above block I wanted to exception handle divide by zero if I wanted to handle a different exception I would have also If I wanted to add a new exception like value error all I have to add except keyword and that exception

```python
try:
    # Any unhandle exception will be under except
    res = "100" / 20
except ArithmeticError:
    print("Arithmetic problem.")
except:
    print("Something went wrong!")       
```

So the difference between all of this incase of 

- try:
- except: (Except catches all error).

if you want to define exceptions like ValueError or anyother
- except ArithmeticError.

for else statement, the try block contains the edge case code if there are not issues it will go to the else block

for finally block, it will always run no matter what the try block did or not.
